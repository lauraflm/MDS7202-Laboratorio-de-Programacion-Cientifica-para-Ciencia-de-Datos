````python
# 1. Imports
import pandas as pd
import numpy as np
import pyarrow.parquet as pq
from pathlib import Path
import matplotlib.pyplot as plt
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score

# 2. Paths correctos

ROOT = Path.cwd()

RAW_DIR      = ROOT / "data" / "raw"
BATCHES_DIR  = ROOT / "data" / "batches"

# subir 2 niveles: airflow → entrega_2 → MDS7202-entrega3
PRED_DIR     = ROOT.parent.parent / "entrega_3" / "predictions"

print("RAW_DIR     :", RAW_DIR)
print("BATCHES_DIR :", BATCHES_DIR)
print("PRED_DIR    :", PRED_DIR)

# 3. Cargar clientes y productos

clientes_df = pq.read_table(RAW_DIR / "clientes.parquet").to_pandas()
productos_df = pq.read_table(RAW_DIR / "productos.parquet").to_pandas()

universe_size = len(clientes_df) * len(productos_df)

print("RAW_DIR      :", RAW_DIR)
print("BATCHES_DIR  :", BATCHES_DIR)
print("PRED_DIR     :", PRED_DIR)
print("N clientes   :", len(clientes_df))
print("N productos  :", len(productos_df))
print("Universo     :", universe_size)
````

````python
def get_batch_performance(batch_id, verbose=True):
    """
    Calcula TP, FP, FN, TN y métricas para un batch dado.
    """

    # 1. Cargar datos 
    real_path = BATCHES_DIR / f"batch_t{batch_id}.parquet"
    real = pq.read_table(real_path).to_pandas()

    real_pairs = (
        real.groupby(["customer_id", "product_id"])
            .size()
            .reset_index()[["customer_id", "product_id"]]
    )
    true_pairs = set(map(tuple, real_pairs.values))

    # 2. Cargar predicciones
    pred_path = PRED_DIR / f"batch_t{batch_id}.csv"
    preds = pd.read_csv(pred_path, names=["customer_id", "product_id"])
    pred_pairs = set(map(tuple, preds.values))

    # 3. Matriz de confusión
    tp = len(pred_pairs & true_pairs)
    fp = len(pred_pairs - true_pairs)
    fn = len(true_pairs - pred_pairs)
    tn = universe_size - tp - fp - fn

    # 4. Métricas
    y_true = [1]*tp + [0]*fp + [1]*fn + [0]*tn
    y_pred = [1]*tp + [1]*fp + [0]*fn + [0]*tn

    metrics = {
        "batch": batch_id,
        "tp": tp,
        "fp": fp,
        "fn": fn,
        "tn": tn,
        "accuracy": accuracy_score(y_true, y_pred),
        "precision": precision_score(y_true, y_pred),
        "recall": recall_score(y_true, y_pred),
        "f1": f1_score(y_true, y_pred),
        "n_predicciones": len(pred_pairs),
        "n_reales": len(true_pairs),
    }

    if verbose:
        print(f"\n=== Batch t{batch_id} ===")
        for k,v in metrics.items():
            print(f"{k}: {v}")

    return metrics
````

#### Detección de Data Drift

¿Detectaron algún cambio significativo (drift) en la distribución de los datos? ¿Cómo lo identificaron?  

Siguiendo con el análisis anterior, tal como se mencionó,  se detectó un cambio significativo en el comportamiento de los datos entre los batches t2 y t3. Desde el punto de vista del negocio, este cambio se manifiesta en un aumento importante de la tasa de compras reales, lo que se refleja en un salto muy notorio del recall. Sin embargo, este incremento viene acompañado de una caída pronunciada de la precision, indicando un aumento considerable en las recomendaciones.

A lo largo de la ejecución del pipeline se observaron señales claras de este cambio. Si bien el pipeline cuenta con una etapa explícita de detección de drift (check_data_drift) que se ejecuta de forma regular en Airflow, el análisis de los artefactos de salida del sistema evidenció un cambio significativo en la forma en que el modelo operó sobre los nuevos datos.

Para complementar, es posible mencionar que, para monitorear data drift implementamos una función detect_drift que compara la media de las variables numéricas entre el set de entrenamiento y las últimas 4 semanas de datos. Consideramos que existe drift cuando una variable presenta un cambio mayor a 0.5 desviaciones estándar (z-score > 0.5) y declaramos drift global cuando al menos 2 columnas superan ese umbral. Entonces para los batches este criterio no fue superado. 

Por lo tanto, este cambio se identificó principalmente a través de indicadores no directos de drift, reflejados en las métricas de desempeño y en los outputs del modelo. Estos cambios simultáneos sugieren que la distribución de los datos de evaluación difiere de la utilizada durante el entrenamiento inicial, dando lugar a un fenómeno de drift que impacta directamente la efectividad del modelo, a pesar de no haber sido detectado por la métrica implementada.

#### Reentrenamiento del modelo

¿Tuvieron que reentrenar su modelo con los nuevos datos? ¿Porqué? ¿Ayudó esto al performance de su modelo?

A pesar del cambio en la distribución de los datos (drift) en el batch t3 y t4 no se realizó reentrenamiento, es decir, la lógica de detección de drift del pipeline no activó dichas etapas. Por lo tanto, el modelo continuó operando con parámetros entrenados previamente, a pesar de que su desempeño no mejorara.

En ambos casos, se ejecutó la rama skip_train del pipeline, reutilizando el modelo entrenado originalmente. En consecuencia, las métricas de desempeño empeoraron, especialmente en términos de F1-score, que alcanzó su valor más bajo en el batch t4. Esto nos indica que el modelo no se adaptó adecuadamente a los cambios en la distribución de los datos, lo que afectó negativamente su capacidad para realizar predicciones precisas. En este sentido, el comportamiento observado indica que, si bien el reentrenamiento automático es una buena herramienta, su efectividad depende fuertemente de la métrica de drift y del umbral utilizados, los cuales para este caso resultaron ser poco relevantes (sensibles) a los cambios reales en los datos.

````python
import joblib
import sys, types

# Crear módulo dummy
dummy = types.ModuleType("dummy")

# --- FUNCIONES DUMMY NECESARIAS PARA EL PIPELINE ---
def week_to_sin_cos(x):
    return x

# Clase dummy para IQRClipper
class IQRClipper:
    def __init__(self, *args, **kwargs): pass
    def fit(self, X, y=None): return self
    def transform(self, X): return X

# Registrar funciones y clases dentro del dummy
dummy.week_to_sin_cos = week_to_sin_cos
dummy.IQRClipper = IQRClipper

# Registrar módulos que el pickle busca
sys.modules['src'] = dummy
sys.modules['src.data_io'] = dummy
sys.modules['src.config'] = dummy
sys.modules['src.preprocessing'] = dummy

# Cargar modelo
model = joblib.load("artifacts/models/xgb_best.pkl")

print("Modelo cargado correctamente:")
print(model)

````

````python
from sklearn.compose import ColumnTransformer

pre = model.named_steps["preprocessor"]
xgb = model.named_steps["model"]

feature_names = []

for name, transformer, cols in pre.transformers_:
    if name == "remainder":
        continue
    
    if hasattr(transformer, "named_steps"):
        # pipelines como week, num, cat
        last = transformer.named_steps[list(transformer.named_steps.keys())[-1]]
        if hasattr(last, "get_feature_names_out"):
            fn = last.get_feature_names_out(cols)
        else:
            fn = cols
    elif hasattr(transformer, "get_feature_names_out"):
        fn = transformer.get_feature_names_out(cols)
    else:
        fn = cols
    
    feature_names.extend(fn)

print("N features finales:", len(feature_names))
````

````python
import numpy as np
import pandas as pd

pre = model.named_steps["preprocessor"]
xgb = model.named_steps["model"]

feature_names = []

for name, trans, cols in pre.transformers_:

    # Saltar columnas eliminadas
    if name == "remainder":
        continue

    # --- 1. Pipeline (caso: week, num, cat) ---
    if hasattr(trans, "named_steps"):

        # último paso de ese sub-pipeline
        last_step = list(trans.named_steps.values())[-1]

        # Caso OneHotEncoder -> sí tiene get_feature_names_out()
        if hasattr(last_step, "get_feature_names_out"):
            fn = last_step.get_feature_names_out(cols)

        # Caso FunctionTransformer -> NO tiene names → generamos 2 columnas (sin, cos)
        elif name == "week":
            fn = [f"{cols[0]}_sin", f"{cols[0]}_cos"]

        # Caso transformadores numéricos → mantienen el nombre original
        else:
            fn = cols

        feature_names.extend(fn)
        continue

    # --- 2. Transformador directo con get_feature_names_out ---
    if hasattr(trans, "get_feature_names_out"):
        fn = trans.get_feature_names_out(cols)
        feature_names.extend(fn)
        continue

    # --- 3. Último caso: el transformador no tiene names (usamos nombres originales) ---
    feature_names.extend(cols)


print("Nombres de features generados:", len(feature_names))
print("Nombres vs importancias:", len(feature_names), len(xgb.feature_importances_))
````

#### Limitaciones en el modelo

¿Qué limitaciones detectaron en su modelo o en los datos?  


Una de las principales limitaciones que enfrentamos fue el fuerte desbalance de clases. Solo una fracción muy pequeña de los pares cliente–producto realmente compra, lo que hace que el modelo tienda naturalmente a predecir “no compra”. Esto dificulta que el algoritmo aprenda patrones positivos y obliga a usar ajustes como scale_pos_weight para compensar el desbalance.

Otra limitación importante es el cambio en la distribución de los datos entre semanas. A lo largo de los distintos batches observamos variaciones en la cantidad de compras, tipos de productos vendidos y composición de clientes. Estos cambios afectan el desempeño del modelo, ya que lo que aprende en un batch no siempre se mantiene estable en los siguientes. En la práctica, esto significa que el modelo es sensible al drift y puede perder precisión si no se actualiza regularmente.

También notamos que la información disponible sobre los clientes es limitada. El modelo termina dependiendo mucho de las características del producto (como categoría, formato o marca) y menos del comportamiento individual del cliente. Esto reduce la capacidad del modelo para capturar hábitos más personalizados y limita la posibilidad de hacer predicciones muy específicas por cliente.

Finalmente, el uso de técnicas como OneHotEncoder genera un número muy grande de variables, especialmente para las columnas categóricas. Esto puede introducir ruido, sobre todo cuando existen categorías muy poco frecuentes, y hace que el modelo sea más sensible a datos nuevos o raros.

En conjunto, estas limitaciones muestran que, si bien el modelo funciona razonablemente bien, aún hay oportunidades de mejora en la calidad de los datos, el manejo del drift y la cantidad de información disponible para describir a los clientes y sus hábitos de compra.
