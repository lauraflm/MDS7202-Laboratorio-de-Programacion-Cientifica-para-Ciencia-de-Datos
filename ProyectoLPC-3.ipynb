En el competitivo universo de las bebidas gaseosas, la empresa **SodAI Drinks 🥤** ha logrado destacarse por su creatividad, diversidad de productos y enfoque centrado en el cliente. Ofrece una extensa gama de bebidas carbonatadas que abarca distintos segmentos del mercado: desde productos premium en presentaciones sofisticadas, hasta gaseosas accesibles para el consumo masivo, disponibles en diversos tamaños y tipos de envases.

La compañía opera en múltiples regiones y zonas, sirviendo a una variedad de puntos de venta que incluyen desde tiendas de conveniencia y minimarkets hasta el canal frío tradicional. Cada tipo de cliente tiene sus particularidades: algunos reciben entregas hasta 4 veces por semana, mientras que otros son visitados por la fuerza de ventas solo una vez semanalmente. Esta diversidad de perfiles representa tanto una oportunidad como un desafío comercial: ¿cómo saber qué productos tienen más chances de ser comprados por cada cliente en un momento dado?

Con el objetivo de aumentar la facturación de forma inteligente y mejorar la eficiencia de su estrategia de ventas, **SodAI Drinks** decide crear una nueva célula interna de innovación: el equipo **Deep Drinkers 🤖**, cuyo propósito es aplicar ciencia de datos para anticiparse a las necesidades del cliente y potenciar el negocio desde una perspectiva basada en información.

El corazón de esta iniciativa es el desarrollo de un sistema predictivo personalizado para cada cliente. Para ello, **Deep Drinkers** convoca a un equipo de Data Scientists y especialistas en *machine learning* con una misión clara: construir un modelo predictivo que, cada semana, pueda estimar la probabilidad de compra de cada producto del portafolio para cada cliente activo.

El modelo deberá tener en cuenta múltiples factores, incluyendo:
- **Tipo de cliente**, ej. "TIENDA DE CONVENIENCIA", "MINIMARKET".
- **Frecuencia de entregas y visitas**, indicadores del nivel de actividad comercial.
- **Ubicación geográfica** (por región y zona).
- **Preferencias históricas de consumo**, inferidas por patrones de compra anteriores.
- **Características del producto**, como marca, categoría, segmento, tipo de envase y tamaño

El objetivo final es que, **cada semana**, se genere una tabla de productos priorizados: para cada cliente, un listado de productos ordenado por su probabilidad estimada de compra. Esta información será enviada al equipo comercial, que podrá usarla en call center, para incrementar las chances de concretar ventas al ofrecer justo lo que el cliente probablemente quiere comprar.

Este proyecto representa un cambio de paradigma en la forma en que **SodAI Drinks** gestiona su fuerza de ventas: de un enfoque reactivo y generalista, a uno proactivo, basado en datos y profundamente personalizado. Así, la empresa no solo espera aumentar su rentabilidad, sino también construir relaciones más sólidas con sus clientes, ofreciéndoles recomendaciones más relevantes y oportunas.

Para lograr lo anterior, el equipo **Deep Drinkers** contará con los siguientes conjuntos de datos, junto a sus respectivos atributos:

- **Datos transaccionales** (`transacciones.parquet`): contiene el historial de compras realizadas por los clientes.
	- `customer_id`: identificador único del cliente que realizó la compra.
	- `product_id`: identificador único del producto comprado.
	- `purchase_date`: fecha en que se realizó la transacción.
	- `order_id`: identificar de la orden de su pedido.
	- `payment`	monto total pagado por la transacción.

- **Datos de clientes** (`clientes.parquet`): incluye las características de cada cliente.
	- `customer_id`: identificador único del cliente.
	- `region_id`: identificador de la región geográfica donde se encuentra el cliente.
	- `customer_type`: tipo de cliente según el canal comercial, por ejemplo, “TIENDA DE CONVENIENCIA”.
	- `Y`: coordenada geográfica de latitud.
	- `X`: coordenada geográfica de longitud.
	- `num_deliver_per_week`: cantidad de entregas semanales que recibe el cliente.
	- `num_visit_per_week`: frecuencia de visitas de la fuerza de ventas por semana.

- **Datos de productos** (`productos.parquet`): describe las características de los productos del portafolio.
	- `product_id`: identificador único del producto.
	- `brand`: marca comercial del producto.
	- `category`: categoría general del producto, como “BEBIDAS CARBONATADAS”.
	- `sub_category`: subcategoría dentro de la categoría principal, por ejemplo, “GASEOSAS”.
	- `segment`: segmento de mercado al que pertenece el producto, como “PREMIUM”.
	- `package`: tipo de envase del producto.
	- `size`: tamaño del producto en litros.

Se crea la base del tipo semanal, en donde como se tienen datos de un año completo, deberían quedar 52 semanas en total.

```python
import pandas as pd
import numpy as np

# base semanal SOLO por 'semana'
tx = df_transacciones.copy()
tx['purchase_date'] = pd.to_datetime(tx['purchase_date']).dt.normalize()
tx['semana'] = tx['purchase_date'].dt.isocalendar().week.astype(int)

# par cliente–producto–semana
base_cp_sem = (tx
    .groupby(['customer_id','product_id','semana'], as_index=False)
    .agg(
        items_cp=('items','sum'),
        pedidos_cp=('order_id','nunique')
    )
    .assign(compra_flag=lambda d: (d['items_cp'] > 0).astype('int8'))
)
```

Después, se hace el producto entre clientes, productos y semana, esto con el fin de poder tener un análisis de todas las compras y no compras que se hicieron de cada uno de los productos.

```python
# producto cruz C×P×semana (incluye no-transacciones)
weeks_total     = np.sort(tx['semana'].unique())
customers_total = tx['customer_id'].unique()
products_total  = tx['product_id'].unique()

idx = pd.MultiIndex.from_product(
    [customers_total, products_total, weeks_total],
    names=['customer_id','product_id','semana']
)
full = idx.to_frame(index=False)

print(f"Grid esperado: {len(customers_total)} clientes × {len(products_total)} productos × {len(weeks_total)} semanas = {len(full):,} filas")
```

También se crea la variable "compra_flag" que indica si un cliente compró o no y se crea además, "pedidos_cp" que indica la cantidad de productos que compra.

```python
issues = {}

# Valores negativos o cero donde no corresponde
issues['items_negativos_o_cero'] = df_transacciones.query('items <= 0')
issues['size_negativo_o_cero'] = df_productos.query('size <= 0')

# Fechas fuera de rango (futuras o muy antiguas)
hoy = pd.Timestamp.today().normalize()
issues['fecha_futura'] = df_transacciones[df_transacciones['purchase_date'] > hoy]
issues['fecha_muy_antigua'] = df_transacciones[df_transacciones['purchase_date'] < pd.Timestamp('2000-01-01')]

# IDs inexistentes respecto de maestros
ids_clientes = set(df_clientes['customer_id'])
ids_productos = set(df_productos['product_id'])
issues['cliente_no_en_maestro'] = df_transacciones[~df_transacciones['customer_id'].isin(ids_clientes)]
issues['producto_no_en_maestro'] = df_transacciones[~df_transacciones['product_id'].isin(ids_productos)]

# Duplicados sospechosos (mismo cliente-producto-orden)
issues['duplicados_cliente_prod_orden'] = (
    df_transacciones
    .duplicated(subset=['customer_id','product_id','order_id'], keep=False)
)

# Coordenadas fuera de rango (si X/Y son long/lat aprox. en grados)
issues['coords_fuera_rango'] = df_clientes[
    (~df_clientes['X'].between(-180,180)) | (~df_clientes['Y'].between(-90,90))]
```

```python
# Semana calendario (ISO) desde purchase_date
df_transacciones['week'] = df_transacciones['purchase_date'].dt.isocalendar().week.astype(int)
df_transacciones['year'] = df_transacciones['purchase_date'].dt.year.astype(int)

# Distintos productos comprados por cliente-semana
prods_cliente_semana = (
    df_transacciones
      .groupby(['customer_id','year','week'])['product_id']
      .nunique()
      .rename('n_skus')
      .reset_index()
)

# Promedio por cliente (promedio de sus semanas activas)
prom_prod_sem_cliente = (
    prods_cliente_semana
      .groupby('customer_id')['n_skus']
      .mean()
      .rename('prom_skus_por_semana')
      .reset_index()
)

# Promedio global
promedio_global_skus_semana = prom_prod_sem_cliente['prom_skus_por_semana'].mean()
promedio_global_skus_semana
```

```python
txs_por_cliente = (
    df_transacciones
      .groupby('customer_id')['order_id']
      .nunique()  # si cada order_id es una transacción; usa .size() si cada fila ya es transacción
      .rename('n_transacciones')
      .reset_index())
```

```python
# Ordenamos y calculamos deltas por cliente y SKU
df_sorted = df_transacciones.sort_values(['customer_id','product_id','purchase_date']).copy()
df_sorted['delta_dias'] = (
    df_sorted
      .groupby(['customer_id','product_id'])['purchase_date']
      .diff()
      .dt.days
)

# Quitamos la primera compra (sin delta) y valores raros (negativos o enormes)
valid = df_sorted['delta_dias'].between(1, 365*2)  # por ejemplo, <= 2 años
repurchase_deltas = df_sorted.loc[valid, ['product_id','delta_dias']]

# Promedio de recompra por SKU
periodo_recompra_por_sku = (
    repurchase_deltas
      .groupby('product_id')['delta_dias']
      .mean()
      .round(1)
      .rename('dias_recompra_prom')
      .reset_index()
      .sort_values('dias_recompra_prom')
)

# (Opcional) también puedes calcular la mediana, que suele ser más robusta
periodo_recompra_por_sku_mediana = (
    repurchase_deltas
      .groupby('product_id')['delta_dias']
      .median()
      .rename('dias_recompra_med')
      .reset_index()
)
```

```python
import numpy as np
import pandas as pd
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import OneHotEncoder, OrdinalEncoder, StandardScaler, FunctionTransformer, KBinsDiscretizer
from sklearn.impute import SimpleImputer
from sklearn.base import BaseEstimator, TransformerMixin
```

Entonces, se hace una clase IQR, la cual funciona de buena forma para poder sacar los outliers que se tienen en la base de datos.

Además, se hace la transformación a semana de forma cíclica, para que la distancia que exista entre semanas sea equiparable.

```python
#Recorte IQR para outliers numéricos
class IQRClipper(BaseEstimator, TransformerMixin):
    def __init__(self, factor=1.5):
        self.factor = factor
        self.bounds_ = None
    def fit(self, X, y=None):
        X = pd.DataFrame(X)
        self.bounds_ = []
        for c in X.columns:
            q1, q3 = X[c].quantile([0.25, 0.75])
            iqr = q3 - q1
            lo, hi = q1 - self.factor*iqr, q3 + self.factor*iqr
            self.bounds_.append((lo, hi))
        return self
    def transform(self, X):
        X = pd.DataFrame(X).copy()
        for i, c in enumerate(X.columns):
            lo, hi = self.bounds_[i]
            X[c] = X[c].clip(lo, hi)
        return X.values

# Codificación cíclica para 'semana' (1..52)
def week_to_sin_cos(a):
    a = np.asarray(a).reshape(-1, 1).astype(float)
    P = 52.0
    sin_ = np.sin(2*np.pi*(a/P))
    cos_ = np.cos(2*np.pi*(a/P))
    return np.hstack([sin_, cos_])
```

```python
# OneHotEncoder compatible (sklearn viejas/nuevas)
def make_ohe(min_freq=None):
    try:
        return OneHotEncoder(handle_unknown='ignore', sparse_output=True,
                             **({'min_frequency':min_freq} if min_freq is not None else {}))
    except TypeError:
        return OneHotEncoder(handle_unknown='ignore', sparse=True,
                             **({'min_frequency':min_freq} if min_freq is not None else {}))
```

Se crea una clase personalizada que agrega dos nuevas variables importantes:
1. **compro_semana_pasada**: Indica si el cliente compró algún producto la semana anterior (variable binaria)
2. **promedio_compra**: Promedio histórico de compras del cliente hasta la semana actual

Estas variables capturan el comportamiento de compra reciente y histórico del cliente, lo cual puede ser muy predictivo para determinar la probabilidad de compra futura.

```python
class ComportamientoCompraTransformer(BaseEstimator, TransformerMixin):
    """
    Transformer
    - compro_semana_pasada: Si el cliente compró algún producto la semana anterior
    - promedio_compra: Promedio histórico de compras del cliente hasta la semana actual
    """

    def __init__(self, customer_col='customer_id', semana_col='semana', target_col='compra_flag'):
        self.customer_col = customer_col
        self.semana_col = semana_col
        self.target_col = target_col

    def fit(self, X, y=None):
        return self

    def transform(self, X):
        # dataframe
        if not isinstance(X, pd.DataFrame):
            X = pd.DataFrame(X)

        X_transformed = X.copy()

        # Compro semana pasada
        compras_por_semana = (X_transformed[X_transformed[self.target_col] == 1]
                             .groupby([self.customer_col, self.semana_col])
                             .size()
                             .reset_index(name='compro_flag_temp'))
        compras_por_semana['semana_anterior'] = compras_por_semana[self.semana_col] - 1

        # Merge para saber si compró la semana anterior
        compro_anterior = compras_por_semana[[self.customer_col, 'semana_anterior']].rename(
            columns={'semana_anterior': self.semana_col})
        compro_anterior['compro_semana_pasada'] = 1

        # Merge con el dataset original
        X_transformed = X_transformed.merge(
            compro_anterior[[self.customer_col, self.semana_col, 'compro_semana_pasada']],
            on=[self.customer_col, self.semana_col],
            how='left'
        )
        X_transformed['compro_semana_pasada'] = X_transformed['compro_semana_pasada'].fillna(0).astype(int)

        # Promedio históricoo
        X_sorted = X_transformed.sort_values([self.customer_col, self.semana_col])

        # promedio acumulativo
        X_sorted['promedio_compra'] = (X_sorted.groupby(self.customer_col)[self.target_col]
                                      .expanding()
                                      .mean()
                                      .shift(1)  # Shift para no incluir la semana actual
                                      .fillna(0)  # Primera semana = 0
                                      .values)

        X_transformed = X_sorted.sort_index()

        return X_transformed

# instacia del transformer
comportamiento_transformer = ComportamientoCompraTransformer(
    customer_col='customer_id',
    semana_col='semana',
    target_col='compra_flag'
)
```

```python
ID_COLS = ['customer_id','product_id']
TARGET   = 'compra_flag'
WEEK     = ['semana']

# Categóricas
CAT_LOW  = [c for c in ['customer_type','sub_category','segment','package'] if c in df.columns]  # OHE
CAT_HIGH = [c for c in ['brand'] if c in df.columns]                                           # Ordinal

# Numéricas base (sin constantes ni duplicadas)

NUM = [c for c in ['size','num_deliver_per_week','X','Y','compro_semana_pasada','promedio_compra'] if c in df.columns]
```

Se hace el pipeline para hacer la transformación a la variable semana, se hace también para las numéricas haciendo un StandarScaler, también se realiza para las categóricas con lo explicado anteriormente y se hace una discretización para size.

```python
# Semana -> sin/cos
week_encoder = Pipeline([
    ('imp', SimpleImputer(strategy='median')),
    ('cyc', FunctionTransformer(week_to_sin_cos,
            feature_names_out=lambda tf,n: ['semana_sin','semana_cos']))])

# Numéricos con outlier clipping + escala
num_pipe = Pipeline([
    ('imp', SimpleImputer(strategy='median')),
    ('clip', IQRClipper(factor=1.5)),     # recorta X,Y,size,num_deliver_per_week
    ('sc', StandardScaler())])

# Discretización para size (bins por cuantiles, ordinal -> liviano)
USE_SIZE_BIN = True
if USE_SIZE_BIN and 'size' in NUM:
    size_bin_pipe = Pipeline([
        ('imp', SimpleImputer(strategy='median')),
        ('kb',  KBinsDiscretizer(n_bins=4, encode='ordinal', strategy='quantile'))
    ])
    SIZE_BIN_COLS = ['size']
else:
    SIZE_BIN_COLS = []

# OHE para baja cardinalidad
ohe_low = Pipeline([
    ('imp', SimpleImputer(strategy='constant', fill_value='UNK')),
    ('ohe', make_ohe(min_freq=None))])

# Ordinal para alta cardinalidad
ord_high = Pipeline([
    ('imp', SimpleImputer(strategy='constant', fill_value='UNK')),
    ('ord', OrdinalEncoder(handle_unknown='use_encoded_value', unknown_value=-1))])
```

Ahora se generan los transformers:

```python
transformers = [
    ('week', week_encoder, WEEK),
    ('num',  num_pipe,    NUM),]

if SIZE_BIN_COLS:
    transformers.append(('sizebin', size_bin_pipe, SIZE_BIN_COLS))
if CAT_LOW:
    transformers.append(('ohe',  ohe_low,  CAT_LOW))
if CAT_HIGH:
    transformers.append(('ord',  ord_high, CAT_HIGH))

preprocessor_cls = ColumnTransformer(
    transformers=transformers,
    remainder='drop',
    sparse_threshold=1.0)  # mantiene salida esparsa si es posible
```

```python
# Definiciones previas
USE_SIZE_BIN = True  # True si quieres KBins sobre 'size'

WEEK = ['semana']
CAT_LOW  = [c for c in ['customer_type','sub_category','segment','package'] if c in df.columns]
CAT_HIGH = [c for c in ['brand'] if c in df.columns]

# Numéricas base
NUM_BASE = [c for c in ['size','num_deliver_per_week','X','Y','compro_semana_pasada','promedio_compra'] if c in df.columns]

# Si discretizamos 'size', la sacamos de los numéricos continuos
if USE_SIZE_BIN and 'size' in NUM_BASE:
    NUM = [c for c in NUM_BASE if c != 'size']
    SIZE_BIN_COLS = ['size']
else:
    NUM = NUM_BASE
    SIZE_BIN_COLS = []

# ver grupos
print("WEEK:", WEEK)
print("NUM :", NUM)
print("SIZE_BIN_COLS:", SIZE_BIN_COLS)
print("CAT_LOW:", CAT_LOW)
print("CAT_HIGH:", CAT_HIGH)


_sel = WEEK + NUM + SIZE_BIN_COLS + CAT_LOW + CAT_HIGH
_dups = [c for c in _sel if _sel.count(c) > 1]
assert not _dups, f"Columnas repetidas entre transformers: {sorted(set(_dups))}"
```

```python
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import OneHotEncoder, OrdinalEncoder, StandardScaler, FunctionTransformer, KBinsDiscretizer
from sklearn.impute import SimpleImputer


#Definidos antes: week_encoder, num_pipe, make_ohe, etc. —
def week_to_sin_cos(a):
    a = np.asarray(a).reshape(-1, 1).astype(float)
    P = 52.0
    return np.hstack([np.sin(2*np.pi*(a/P)), np.cos(2*np.pi*(a/P))])

week_encoder = Pipeline([
    ('imp', SimpleImputer(strategy='median')),
    ('cyc', FunctionTransformer(week_to_sin_cos,
            feature_names_out=lambda tf,n: ['semana_sin','semana_cos']))])

num_pipe = Pipeline([
    ('imp', SimpleImputer(strategy='median')),
    ('sc', StandardScaler())])

def make_ohe(min_freq=None):
    try:
        return OneHotEncoder(handle_unknown='ignore', sparse_output=True,
                             **({'min_frequency':min_freq} if min_freq is not None else {}))
    except TypeError:
        return OneHotEncoder(handle_unknown='ignore', sparse=True,
                             **({'min_frequency':min_freq} if min_freq is not None else {}))

ohe_low = Pipeline([
    ('imp', SimpleImputer(strategy='constant', fill_value='UNK')),
    ('ohe', make_ohe(min_freq=None))
])

ord_high = Pipeline([
    ('imp', SimpleImputer(strategy='constant', fill_value='UNK')),
    ('ord', OrdinalEncoder(handle_unknown='use_encoded_value', unknown_value=-1))
])

# Discretización de 'size' (solo si está activada)
size_bin_pipe = Pipeline([
    ('imp', SimpleImputer(strategy='median')),
    ('kb',  KBinsDiscretizer(n_bins=4, encode='ordinal', strategy='quantile'))
]) if USE_SIZE_BIN and SIZE_BIN_COLS else None

# Armar ColumnTransformer sin columnas repetidas
transformers = [
    ('week', week_encoder, WEEK),
    ('num',  num_pipe,    NUM),]

if size_bin_pipe:
    transformers.append(('sizebin', size_bin_pipe, SIZE_BIN_COLS))
if CAT_LOW:
    transformers.append(('ohe',  ohe_low,  CAT_LOW))
if CAT_HIGH:
    transformers.append(('ord',  ord_high, CAT_HIGH))

preprocessor_cls = ColumnTransformer(
    transformers=transformers,
    remainder='drop',
    sparse_threshold=1.0)
```

```python
# Orden temporal y split ya creados antes
df = df.sort_values(['customer_id','product_id','semana'])

TARGET = 'compra_flag'
X_cols = WEEK + NUM + SIZE_BIN_COLS + CAT_LOW + CAT_HIGH  # <- coherente con 'transformers'
y_col  = TARGET

X_train = df.loc[df['split']=='train', X_cols]
y_train = df.loc[df['split']=='train', y_col].astype('int8')

X_val   = df.loc[df['split']=='val',   X_cols]
y_val   = df.loc[df['split']=='val',   y_col].astype('int8')

X_test  = df.loc[df['split']=='test',  X_cols]
y_test  = df.loc[df['split']=='test',  y_col].astype('int8')

preprocessor_cls.fit(X_train)
Xtr = preprocessor_cls.transform(X_train)  # no .toarray()
Xva = preprocessor_cls.transform(X_val)
Xte = preprocessor_cls.transform(X_test)

print("Shapes ->", Xtr.shape, Xva.shape, Xte.shape)
```

```python
from sklearn.metrics import (
    classification_report,
    confusion_matrix,
    roc_auc_score,
    average_precision_score)

def _get_scores(model, X):
    # Probabilidades si existen; si no, usa decision_function y lo escala
    try:
        s = model.predict_proba(X)[:, 1]
    except Exception:
        try:
            d = model.decision_function(X)
            # Escalado min-max para aproximar a [0,1]
            dmin, dmax = np.min(d), np.max(d)
            s = (d - dmin) / (dmax - dmin + 1e-12)
        except Exception:
            s = None
    return s

def evaluar_split(nombre, X, y, model):
    y_pred = model.predict(X)
    y_score = _get_scores(model, X)

    print(f"\n {nombre} ")
    print("Distribución real (0/1):", np.bincount(y.astype(int)))
    print(classification_report(y, y_pred, digits=3))

    cm = confusion_matrix(y, y_pred)
    tn, fp, fn, tp = cm.ravel()
    print("Matriz de confusión:\n", cm)
    print(f"Precision(positiva) = {tp/(tp+fp+1e-12):.3f} | Recall(positiva) = {tp/(tp+fn+1e-12):.3f}")

    if y_score is not None:
        try:
            roc = roc_auc_score(y, y_score)
            pr  = average_precision_score(y, y_score)
            print(f"ROC-AUC = {roc:.3f} | PR-AUC = {pr:.3f}")
        except Exception:
            pass
```

```python
# Optimización XGBoost con Optuna - F1 Score clase positiva
import optuna
import xgboost as xgb
import gc
import numpy as np
from sklearn.metrics import f1_score
import warnings
import os, joblib
from sklearn.metrics import roc_auc_score, precision_score, recall_score, confusion_matrix
warnings.filterwarnings('ignore')
optuna.logging.set_verbosity(optuna.logging.WARNING)

def objective(trial):
    # Parámetros XGBoost optimizados
    params = {
        'n_estimators': trial.suggest_int('n_estimators', 50, 300),
        'max_depth': trial.suggest_int('max_depth', 3, 8),
        'learning_rate': trial.suggest_float('learning_rate', 0.01, 0.3, log=True),
        'subsample': trial.suggest_float('subsample', 0.6, 1.0),
        'colsample_bytree': trial.suggest_float('colsample_bytree', 0.6, 1.0),
        'reg_alpha': trial.suggest_float('reg_alpha', 1e-8, 10.0, log=True),
        'reg_lambda': trial.suggest_float('reg_lambda', 1e-8, 10.0, log=True),
        'random_state': 42,
        'verbosity': 0,
        'tree_method': 'hist',
        'n_jobs': -1
    }

    try:
        #scale_pos_weight para el desbalance
        neg_count = (y_train == 0).sum()
        pos_count = (y_train == 1).sum()
        params['scale_pos_weight'] = neg_count / pos_count

        # USAR LOS DATOS TRANSFORMADOS (Xtr, Xva)
        model = xgb.XGBClassifier(**params)
        model.fit(Xtr, y_train)

        # Optimización F1 SCORE DE LA CLASE POSITIVA
        y_pred_proba = model.predict_proba(Xva)[:, 1]
        thresholds = np.linspace(0.1, 0.9, 17)
        best_f1 = 0

        for threshold in thresholds:
            y_pred_binary = (y_pred_proba >= threshold).astype(int)
            f1 = f1_score(y_val, y_pred_binary, pos_label=1)
            if f1 > best_f1:
                best_f1 = f1

        del model, y_pred_proba, y_pred_binary
        gc.collect()
        return best_f1

    except Exception as e:
        print(f"Error en trial {trial.number}: {e}")
        gc.collect()
        return 0.0


study = optuna.create_study(
    direction='maximize',
    pruner=optuna.pruners.MedianPruner(n_startup_trials=5, n_warmup_steps=10)
)

# Configuración de trials
n_trials = 60
timeout = 600

print(f"Configuración: {n_trials} trials, {timeout//60} min timeout")
print(f"Métrica objetivo: F1 Score clase positiva")

study.optimize(objective, n_trials=n_trials, timeout=timeout)

print(f"\nBest F1 Score: {study.best_value:.4f}")
print("Best params:")
for k, v in study.best_params.items():
    if isinstance(v, float):
        print(f"  {k}: {v:.4f}")
    else:
        print(f"  {k}: {v}")

# Entrenar modelo final optimizado
final_params = study.best_params.copy()
final_params.update({
    'random_state': 42,
    'verbosity': 0,
    'tree_method': 'hist',
    'n_jobs': -1
})

t
neg_count = (y_train == 0).sum()
pos_count = (y_train == 1).sum()
final_params['scale_pos_weight'] = neg_count / pos_count

best_model = xgb.XGBClassifier(**final_params)

# USAR LOS DATOS TRANSFORMADOS
best_model.fit(Xtr, y_train)

joblib.dump(best_model, "xgb_best.pkl")

# Evaluación
def evaluate_with_best_threshold(model, X, y, split_name):
    y_pred_proba = model.predict_proba(X)[:, 1]

    # Encontrar mejor threshold para F1 de clase positiva
    thresholds = np.linspace(0.1, 0.9, 21)
    best_f1 = 0
    best_threshold = 0.5

    for threshold in thresholds:
        y_pred_binary = (y_pred_proba >= threshold).astype(int)
        f1 = f1_score(y, y_pred_binary, pos_label=1)
        if f1 > best_f1:
            best_f1 = f1
            best_threshold = threshold


    y_pred_final = (y_pred_proba >= best_threshold).astype(int)

    # Calcular todas las métricas
    f1_final = f1_score(y, y_pred_final, pos_label=1)
    roc_auc = roc_auc_score(y, y_pred_proba)
    precision = precision_score(y, y_pred_final, pos_label=1)
    recall = recall_score(y, y_pred_final, pos_label=1)

    return {
        'f1_score': f1_final,
        'roc_auc': roc_auc,
        'precision': precision,
        'recall': recall,
        'best_threshold': best_threshold,
        'predictions': y_pred_final,
        'probabilities': y_pred_proba
    }

# evaluar en validation y test
val_results = evaluate_with_best_threshold(best_model, Xva, y_val, "Validation")
test_results = evaluate_with_best_threshold(best_model, Xte, y_test, "Test")

print(f"\nResultados finales:")
print(f"  Validation F1 Score: {val_results['f1_score']:.4f} (threshold: {val_results['best_threshold']:.2f})")
print(f"  Validation ROC-AUC: {val_results['roc_auc']:.4f}")
print(f"  Test F1 Score: {test_results['f1_score']:.4f} (threshold: {test_results['best_threshold']:.2f})")
print(f"  Test ROC-AUC: {test_results['roc_auc']:.4f}")

# Matriz de confusión
cm_val = confusion_matrix(y_val, val_results['predictions'])
print(f"\nMétricas detalladas (Validation):")
print(f"  Precision: {val_results['precision']:.4f}")
print(f"  Recall: {val_results['recall']:.4f}")
print(f"  F1 Score: {val_results['f1_score']:.4f}")
print(f"  Matriz de confusión:")
print(f"  {cm_val}")

gc.collect()
```
